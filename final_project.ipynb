```python
import csv
import matplotlib.pyplot as plt
from datetime import datetime
import os
Filename  = "expenses.csv"
```

```python
budget_goals = {}


def set_budget():
  category = input("creat a category to set a budget for").strip().lower()
  Goal = float(input(f"Set your budget for '{category}'"))
  budget_goals[category] = Goal
  print(f"budget set for '{category}, ${Goal}\n'")

```

```python
def add_expense():
  amount= float(input("Enter the amount of the expense $$$"))
  Category= input("What category (ex. rent,bills,groceries)").strip().lower()
  Date = input("Enter the date (year-month-day)")

  with open(Filename, mode="a",newline="") as file:
      writer = csv.writer(file)
      writer.writerow([amount,Category,Date])

  print(f"expense of ${amount} added to '{Category}' on {Date}.\n")

```

```python
def read_expenses():
  expenses = []
  if os.path.exists(Filename):
      with open(Filename, mode="r") as file:
          reader = csv.reader(file)
          for row in reader:
              amount = float(row[0])
              category = row[1]
              date = datetime(row[2], "%Y-%m-%d")
              expenses.append({"amount" : amount, "category" : category, "date" : date})
              return expenses
```

```python
def display_summary(expenses):
  totals = {}

  for item in expenses:
    #cat = category
    cat = item["category"]
    totals[cat] = totals.get(cat, 0) + item[amount]

  print("Expense Summary")
  for cat, total in totals.item():
      print(f"{cat.title()}, $(total:.2f)")
  print()

  for cat, goal in budget_goals.items():
      if cat in totals:
          spent = totals[cat]
          if spent > goal:
              print(f"over budget in '{cat}' by ${spent - goal:.2f}")
          else:
             print(f"within budget for '{cat}' (${goal - spent:2f} remaining)")
      else:
         print(f"no spending recorded for budget category '{cat}'")
  print()
```

```python
def plot_barchart(expenses):
    totals = {}
    for item in expenses:
        cat = item["category"]
        totals[cat] = totals.get(cat, 0) + item["amount"]

    categories = list(totals.keys())
    amounts = list(totals.values())

    plt.figure(figsize=(7,4))
    plt.bar(categories, amounts, color="skyblue")
    plt.title("Total Spending By Category")
    plt.xlabel("category")
    plt.ylabel("amount ($)")
    plt.tight_layout()
    plt.show()
```

```python
def plot_trend(expenses):
    trends = {}
    for item in expenses:
        date = item["date"]("year-month")
        cat = item["category"]
        key = (cat, date)
        trends[key] = trends.get(key, 0) + item["amount"]


    categories = set([key[0] for key in trends.keys()])
    dates = sorted(set(key[1] for key in trends.key()))


    for cat in categories:
        y_values = []
        for d in dates:
            y_values.append(trends.get((cat, d), 0))
        plt.plot(dates, y_values, label=cat)


    plt.tile("Spending Trend Over Time")
    plt.xlabel("Date(year-month)")
    plt.ylabel("Amount ($)")
    plt.legend()
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```

```python
def main():
    print("Welcome To Your Personal Budget Tracker!\n")


    while True:
        print("Choose an option")
        print("1. Add Expense")
        print("2. Set Budget Goal")
        print("3. Show Summary")
        print("4. Show Bar Chart")
        print("5. Show Spending Trend")
        print("6. Exit")


        choice = input("Enter your choice (1-6)").strip()
        print()

        expenses = read_expenses()

        if choice == "1":
            add_expense()
        elif choice == "2":
            set_budget()
        elif choice == "3":
            display_summary(expenses)
        elif choice == "4":
            plot_barchart(expenses)
        elif choice == "5":
            plot_trend(expenses)
        elif choice == "6":
            print("Goodbye, Stay on budget")
            break

        else:
            print("Invalid choice. Please enter a number from 1 to 6\n")


main()
```
